```python

import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("/content/seasonal_agriculture_performance_dataset.csv")

df.head()
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
df.columns.tolist()
df.info()
df.isnull().sum()
print("Duplicate rows:", df.duplicated().sum())
# ==========================================
# Step 7: Statistical Summary
# ==========================================

df.describe().T
# Seasonal Agriculture Performance Analysis

## Major Project – VOIS AICTE Batch 1 2026–2027

##Student Name:** Yashika Upadhyay

##Project Domain:** Data Analysis and Visualization

##Tools Used:** Python, Pandas, NumPy, Matplotlib, Seaborn, SciPy, Google Colab

## Project Overview

##This project analyzes agricultural performance across different seasons, crops, environmental conditions, resource usage, irrigation methods, and economic factors.

##The main purpose of this project is to understand seasonal variations in agricultural performance and identify useful patterns and relationships in the dataset.

# ==========================================
# Step 8: Categorical Variables
# ==========================================

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

for column in categorical_columns:
    print("\n" + "=" * 60)
    print(f"Column: {column}")
    print("=" * 60)
    print(df[column].value_counts(dropna=False))
    # ==========================================
# Step 9: Numerical Variables
# ==========================================

numerical_columns = df.select_dtypes(
    include=np.number
).columns

print("Numerical columns:")
print()

for i, column in enumerate(numerical_columns, start=1):
    print(f"{i}. {column}")
# ==========================================
# STEP 2: DATA CLEANING
# 2.1 Create a copy of original dataset
# ==========================================

clean_df = df.copy()

print("Original dataset shape:", clean_df.shape)
# ==========================================
# 2.2 Duplicate Records
# ==========================================

duplicate_count = clean_df.duplicated().sum()

print("Duplicate records found:", duplicate_count)

clean_df = clean_df.drop_duplicates()

print("Dataset shape after removing duplicates:", clean_df.shape)
# ==========================================
# 2.3 Missing Value Check
# ==========================================

missing = clean_df.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

print("Columns containing missing values:")
print(missing)
# ==========================================
# 2.4 Handle Missing Environmental Values
# ==========================================

environmental_columns = [
    "Rainfall_mm",
    "Soil_Moisture_pct"
]

for column in environmental_columns:
    median_value = clean_df[column].median()
    clean_df[column] = clean_df[column].fillna(median_value)

print("Missing environmental values handled.")
# ==========================================
# 2.5 Handle Missing Yield
# ==========================================

before_yield_cleaning = len(clean_df)

clean_df = clean_df.dropna(
    subset=["Yield_Tonnes_Ha"]
)

after_yield_cleaning = len(clean_df)

print("Rows before removing missing Yield:", before_yield_cleaning)
print("Rows after removing missing Yield:", after_yield_cleaning)
print("Rows removed:", before_yield_cleaning - after_yield_cleaning)
# ==========================================
# 2.6 Verify Missing Values
# ==========================================

remaining_missing = clean_df.isnull().sum()

remaining_missing = remaining_missing[
    remaining_missing > 0
]

if len(remaining_missing) == 0:
    print("No missing values remaining.")
else:
    print(remaining_missing)
    # ==========================================
# 2.7 Check Negative Values
# ==========================================

non_negative_columns = [
    "Farm_Area_Hectares",
    "Rainfall_mm",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_Moisture_pct",
    "Nitrogen_kg_ha",
    "Phosphorus_kg_ha",
    "Potassium_kg_ha",
    "Fertilizer_kg_ha",
    "Pesticide_Litre_ha",
    "Seed_Quality_Score",
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Market_Price_INR_Tonne",
    "Total_Cost_INR",
    "Water_Used_m3",
    "Water_Efficiency_t_per_1000m3",
    "Disease_Pest_Risk_pct"
]

for column in non_negative_columns:
    if column in clean_df.columns:
        negative_count = (clean_df[column] < 0).sum()
        print(f"{column}: {negative_count} negative values")
        # ==========================================
# 2.8 Data Type Check
# ==========================================

print(clean_df.dtypes)
# ==========================================
# 2.9 FINAL DATA QUALITY REPORT
# ==========================================

print("=" * 60)
print("FINAL DATA QUALITY REPORT")
print("=" * 60)

print("\nOriginal shape:", df.shape)
print("Cleaned shape :", clean_df.shape)

print("\nDuplicate rows:", clean_df.duplicated().sum())

print("\nTotal missing values:",
      clean_df.isnull().sum().sum())

print("\nNumerical columns:",
      len(clean_df.select_dtypes(include=np.number).columns))

print("\nCategorical columns:",
      len(clean_df.select_dtypes(include=["object", "category"]).columns))
# ==========================================
# 2.10 Save Cleaned Dataset
# ==========================================

cleaned_file = "/content/cleaned_agriculture_data.csv"

clean_df.to_csv(
    cleaned_file,
    index=False
)

print("Cleaned dataset saved successfully.")
print(cleaned_file)
# ==========================================
# STEP 3: EXPLORATORY DATA ANALYSIS
# ==========================================

analysis_df = clean_df.copy()

print("Analysis dataset shape:", analysis_df.shape)
# ==========================================
# 3.2 BASIC DATASET SUMMARY
# ==========================================

print("Total Records:", len(analysis_df))
print("Total Columns:", len(analysis_df.columns))

print("\nDataset Information:")
print(analysis_df.info())
# ==========================================
# 3.3 SEASON DISTRIBUTION
# ==========================================

season_counts = analysis_df["Season"].value_counts()

print(season_counts)
plt.figure(figsize=(8, 5))

season_counts.plot(
    kind="bar"
)

plt.title("Number of Agricultural Records by Season")
plt.xlabel("Season")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
# ==========================================
# 3.4 CROP DISTRIBUTION
# ==========================================

crop_counts = analysis_df["Crop"].value_counts()

print(crop_counts)
plt.figure(figsize=(10, 5))

crop_counts.sort_values().plot(
    kind="barh"
)

plt.title("Number of Records by Crop")
plt.xlabel("Number of Records")
plt.ylabel("Crop")
plt.tight_layout()

plt.show()
# ==========================================
# 3.5 STATE DISTRIBUTION
# ==========================================

state_counts = analysis_df["State"].value_counts()

print(state_counts)
plt.figure(figsize=(10, 5))

state_counts.sort_values().plot(
    kind="barh"
)

plt.title("Number of Agricultural Records by State")
plt.xlabel("Number of Records")
plt.ylabel("State")
plt.tight_layout()

plt.show()
# ==========================================
# 3.6 SEASONAL PERFORMANCE SUMMARY
# ==========================================

season_summary = analysis_df.groupby("Season").agg(
    Records=("Season", "size"),
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Average_Production=("Production_Tonnes", "mean"),
    Average_Revenue=("Revenue_INR", "mean"),
    Average_Cost=("Total_Cost_INR", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Water_Used=("Water_Used_m3", "mean"),
    Average_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean"),
    Average_Disease_Pest_Risk=("Disease_Pest_Risk_pct", "mean")
).round(2)

season_summary
season_summary.to_csv(
    "/content/seasonal_performance_summary.csv"
)

print("Seasonal summary saved successfully.")
# ==========================================
# 3.7 AVERAGE YIELD BY SEASON
# ==========================================

season_yield = (
    analysis_df
    .groupby("Season")["Yield_Tonnes_Ha"]
    .mean()
    .sort_values(ascending=False)
)

print(season_yield)
plt.figure(figsize=(8, 5))

season_yield.plot(
    kind="bar"
)

plt.title("Average Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
# ==========================================
# 3.8 AVERAGE PROFIT BY SEASON
# ==========================================

season_profit = (
    analysis_df
    .groupby("Season")["Profit_INR"]
    .mean()
    .sort_values(ascending=False)
)

print(season_profit)
plt.figure(figsize=(8, 5))

season_profit.plot(
    kind="bar"
)

plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
# ==========================================
# 3.9 AVERAGE PRODUCTION BY SEASON
# ==========================================

season_production = (
    analysis_df
    .groupby("Season")["Production_Tonnes"]
    .mean()
    .sort_values(ascending=False)
)

print(season_production)
plt.figure(figsize=(8, 5))

season_production.plot(
    kind="bar"
)

plt.title("Average Production by Season")
plt.xlabel("Season")
plt.ylabel("Average Production (Tonnes)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
# ==========================================
# 3.10 WATER EFFICIENCY BY SEASON
# ==========================================

water_efficiency = (
    analysis_df
    .groupby("Season")["Water_Efficiency_t_per_1000m3"]
    .mean()
    .sort_values(ascending=False)
)

print(water_efficiency)
plt.figure(figsize=(8, 5))

water_efficiency.plot(
    kind="bar"
)

plt.title("Average Water Efficiency by Season")
plt.xlabel("Season")
plt.ylabel("Water Efficiency (t per 1000 m³)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
# ==========================================
# 3.11 DISEASE / PEST RISK BY SEASON
# ==========================================

risk_by_season = (
    analysis_df
    .groupby("Season")["Disease_Pest_Risk_pct"]
    .mean()
    .sort_values(ascending=False)
)

print(risk_by_season)
plt.figure(figsize=(8, 5))

risk_by_season.plot(
    kind="bar"
)

plt.title("Average Disease and Pest Risk by Season")
plt.xlabel("Season")
plt.ylabel("Risk (%)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
# ==========================================
# 3.12 CROP × SEASON YIELD ANALYSIS
# ==========================================

crop_season_yield = analysis_df.pivot_table(
    values="Yield_Tonnes_Ha",
    index="Crop",
    columns="Season",
    aggfunc="mean"
)

crop_season_yield.round(2)
plt.figure(figsize=(10, 6))

sns.heatmap(
    crop_season_yield,
    annot=True,
    fmt=".2f"
)

plt.title("Average Yield by Crop and Season")
plt.xlabel("Season")
plt.ylabel("Crop")
plt.tight_layout()

plt.show()
# ==========================================
# 3.13 CROP-WISE PROFIT
# ==========================================

crop_profit = (
    analysis_df
    .groupby("Crop")["Profit_INR"]
    .mean()
    .sort_values(ascending=False)
)

print(crop_profit)
plt.figure(figsize=(10, 6))

crop_profit.sort_values().plot(
    kind="barh"
)

plt.title("Average Profit by Crop")
plt.xlabel("Average Profit (INR)")
plt.ylabel("Crop")
plt.tight_layout()

plt.show()
# ==========================================
# 3.14 RAINFALL VS YIELD
# ==========================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=analysis_df,
    x="Rainfall_mm",
    y="Yield_Tonnes_Ha",
    hue="Season",
    alpha=0.7
)

plt.title("Rainfall vs Crop Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()

plt.show()
# ==========================================
# 3.15 WATER USAGE VS YIELD
# ==========================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=analysis_df,
    x="Water_Used_m3",
    y="Yield_Tonnes_Ha",
    hue="Season",
    alpha=0.7
)

plt.title("Water Usage vs Crop Yield")
plt.xlabel("Water Used (m³)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()

plt.show()
# ==========================================
# STEP 4.1 - ENVIRONMENTAL ANALYSIS
# ==========================================

environmental_summary = analysis_df.groupby("Season").agg(
    Average_Rainfall=("Rainfall_mm", "mean"),
    Average_Temperature=("Avg_Temperature_C", "mean"),
    Average_Humidity=("Humidity_pct", "mean"),
    Average_Sunlight=("Sunlight_Hours_Day", "mean"),
    Average_Soil_pH=("Soil_pH", "mean"),
    Average_Soil_Moisture=("Soil_Moisture_pct", "mean")
).round(2)

print("Environmental Conditions by Season")
print("-----------------------------------")

display(environmental_summary)
# ==========================================
# STEP 4.2 - ENVIRONMENTAL CONDITIONS CHART
# ==========================================

environmental_summary.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Environmental Conditions by Season")
plt.xlabel("Season")
plt.ylabel("Average Value")
plt.xticks(rotation=0)
plt.legend(title="Environmental Variables")
plt.tight_layout()

plt.show()
# ==========================================
# STEP 4.2 - NPK, FERTILIZER & PESTICIDE
# ==========================================

nutrient_summary = analysis_df.groupby("Season").agg(
    Average_Nitrogen=("Nitrogen_kg_ha", "mean"),
    Average_Phosphorus=("Phosphorus_kg_ha", "mean"),
    Average_Potassium=("Potassium_kg_ha", "mean"),
    Average_Fertilizer=("Fertilizer_kg_ha", "mean"),
    Average_Pesticide=("Pesticide_Litre_ha", "mean")
).round(2)

print("NPK, Fertilizer and Pesticide Usage by Season")
print("-----------------------------------------------")

display(nutrient_summary)
# ==========================================
# NPK USAGE BY SEASON
# ==========================================

nutrient_summary[
    [
        "Average_Nitrogen",
        "Average_Phosphorus",
        "Average_Potassium"
    ]
].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Average NPK Usage by Season")
plt.xlabel("Season")
plt.ylabel("Average Usage (kg/ha)")
plt.xticks(rotation=0)

plt.legend(
    ["Nitrogen", "Phosphorus", "Potassium"],
    title="Nutrients"
)

plt.tight_layout()
plt.show()
# ==========================================
# FERTILIZER & PESTICIDE USAGE
# ==========================================

nutrient_summary[
    [
        "Average_Fertilizer",
        "Average_Pesticide"
    ]
].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Average Fertilizer and Pesticide Usage by Season")
plt.xlabel("Season")
plt.ylabel("Average Usage")
plt.xticks(rotation=0)

plt.legend(
    ["Fertilizer", "Pesticide"],
    title="Input Type"
)

plt.tight_layout()
plt.show()
# ==========================================
# STEP 4.3 - WATER USAGE & EFFICIENCY
# ==========================================

water_summary = analysis_df.groupby("Season").agg(
    Average_Water_Used=("Water_Used_m3", "mean"),
    Average_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean")
).round(2)

print("Water Usage and Efficiency by Season")
print("-------------------------------------")

display(water_summary)
# ==========================================
# WATER EFFICIENCY BY SEASON
# ==========================================

water_summary["Average_Water_Efficiency"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Water Efficiency by Season")
plt.xlabel("Season")
plt.ylabel("Water Efficiency (t per 1000 m³)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
# ==========================================
# STEP 4.4 - IRRIGATION METHOD ANALYSIS
# ==========================================

print("Number of Records by Irrigation Method")
print("---------------------------------------")

irrigation_counts = analysis_df["Irrigation_Method"].value_counts()

display(irrigation_counts)
# ==========================================
# IRRIGATION PERFORMANCE SUMMARY
# ==========================================

irrigation_summary = analysis_df.groupby(
    "Irrigation_Method"
).agg(
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Average_Production=("Production_Tonnes", "mean"),
    Average_Water_Used=("Water_Used_m3", "mean"),
    Average_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean"),
    Average_Profit=("Profit_INR", "mean")
).round(2)

print("Performance by Irrigation Method")
print("---------------------------------")

display(irrigation_summary)
# ==========================================
# WATER EFFICIENCY BY IRRIGATION METHOD
# ==========================================

irrigation_summary[
    "Average_Water_Efficiency"
].sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Water Efficiency by Irrigation Method")
plt.xlabel("Water Efficiency (t per 1000 m³)")
plt.ylabel("Irrigation Method")

plt.tight_layout()
plt.show()
# ==========================================
# YIELD BY IRRIGATION METHOD
# ==========================================

irrigation_summary[
    "Average_Yield"
].sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Average Yield by Irrigation Method")
plt.xlabel("Average Yield (Tonnes/Ha)")
plt.ylabel("Irrigation Method")

plt.tight_layout()
plt.show()
# ==========================================
# STEP 4.5 - ECONOMIC ANALYSIS
# ==========================================

economic_summary = analysis_df.groupby("Season").agg(
    Average_Revenue=("Revenue_INR", "mean"),
    Average_Cost=("Total_Cost_INR", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Market_Price=("Market_Price_INR_Tonne", "mean")
).round(2)

print("Economic Performance by Season")
print("------------------------------")

display(economic_summary)
# ==========================================
# ECONOMIC PERFORMANCE BY SEASON
# ==========================================

economic_summary[
    [
        "Average_Revenue",
        "Average_Cost",
        "Average_Profit"
    ]
].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Economic Performance by Season")
plt.xlabel("Season")
plt.ylabel("Average Value (INR)")
plt.xticks(rotation=0)

plt.legend(
    ["Revenue", "Cost", "Profit"],
    title="Economic Measures"
)

plt.tight_layout()
plt.show()
# ==========================================
# CROP-WISE ECONOMIC ANALYSIS
# ==========================================

crop_economic = analysis_df.groupby("Crop").agg(
    Average_Revenue=("Revenue_INR", "mean"),
    Average_Cost=("Total_Cost_INR", "mean"),
    Average_Profit=("Profit_INR", "mean")
).round(2)

print("Economic Performance by Crop")
print("----------------------------")

display(crop_economic)
# ==========================================
# AVERAGE PROFIT BY CROP
# ==========================================

crop_economic[
    "Average_Profit"
].sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Average Profit by Crop")
plt.xlabel("Average Profit (INR)")
plt.ylabel("Crop")

plt.tight_layout()
plt.show()
# ==========================================
# AVERAGE PROFIT BY CROP
# ==========================================

crop_economic[
    "Average_Profit"
].sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Average Profit by Crop")
plt.xlabel("Average Profit (INR)")
plt.ylabel("Crop")

plt.tight_layout()
plt.show()
# ============================================================
# STEP 5 - CORRELATION ANALYSIS
# Cell 1 to Cell 6 Combined
# ============================================================

# ------------------------------------------------------------
# CELL 1 - CORRELATION MATRIX
# ------------------------------------------------------------

numeric_df = analysis_df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

print("1. CORRELATION MATRIX")
print("=====================")

display(correlation_matrix.round(2))


# ------------------------------------------------------------
# CELL 2 - CORRELATION HEATMAP
# ------------------------------------------------------------

print("\n2. CORRELATION HEATMAP")
print("======================")

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix of Agricultural Variables")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# CELL 3 - YIELD CORRELATION
# ------------------------------------------------------------

yield_correlations = (
    correlation_matrix["Yield_Tonnes_Ha"]
    .sort_values(ascending=False)
)

print("\n3. CORRELATION WITH CROP YIELD")
print("===============================")

display(yield_correlations.round(3))


# ------------------------------------------------------------
# CELL 4 - YIELD CORRELATION CHART
# ------------------------------------------------------------

print("\n4. YIELD CORRELATION CHART")
print("==========================")

yield_correlations.drop(
    "Yield_Tonnes_Ha"
).sort_values().plot(
    kind="barh",
    figsize=(10, 8)
)

plt.title("Correlation of Agricultural Variables with Crop Yield")
plt.xlabel("Correlation Coefficient")
plt.ylabel("Agricultural Variables")

plt.axvline(
    x=0,
    linestyle="--"
)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# CELL 5 - TOP POSITIVE & NEGATIVE RELATIONSHIPS
# ------------------------------------------------------------

yield_without_self = yield_correlations.drop(
    "Yield_Tonnes_Ha"
)

top_positive = yield_without_self.sort_values(
    ascending=False
).head(5)

top_negative = yield_without_self.sort_values(
    ascending=True
).head(5)

print("\n5. TOP 5 POSITIVE RELATIONSHIPS WITH YIELD")
print("==========================================")

display(top_positive.round(3))

print("\nTOP 5 NEGATIVE RELATIONSHIPS WITH YIELD")
print("========================================")

display(top_negative.round(3))


# ------------------------------------------------------------
# CELL 6 - CROP-WISE YIELD ANALYSIS
# ------------------------------------------------------------

crop_yield_summary = analysis_df.groupby(
    "Crop"
)["Yield_Tonnes_Ha"].agg(
    ["count", "mean", "min", "max"]
).round(2)

print("\n6. YIELD DISTRIBUTION BY CROP")
print("=============================")

display(crop_yield_summary)


# ------------------------------------------------------------
# CROP-WISE YIELD CHART
# ------------------------------------------------------------

print("\nAVERAGE YIELD BY CROP")
print("=====================")

crop_yield_summary["mean"].sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Average Yield by Crop")
plt.xlabel("Average Yield (Tonnes/Ha)")
plt.ylabel("Crop")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# FINAL MESSAGE
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("STEP 5 - CORRELATION ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 55)
# ============================================================
# STEP 6 - STATISTICAL TESTING
# KRUSKAL-WALLIS TEST
# ============================================================

from scipy.stats import kruskal

# ------------------------------------------------------------
# 6.1 - SEASON-WISE YIELD DATA
# ------------------------------------------------------------

kharif_yield = analysis_df[
    analysis_df["Season"] == "Kharif"
]["Yield_Tonnes_Ha"].dropna()

rabi_yield = analysis_df[
    analysis_df["Season"] == "Rabi"
]["Yield_Tonnes_Ha"].dropna()

zaid_yield = analysis_df[
    analysis_df["Season"] == "Zaid"
]["Yield_Tonnes_Ha"].dropna()


# ------------------------------------------------------------
# 6.2 - DISPLAY SAMPLE COUNTS
# ------------------------------------------------------------

print("SEASON-WISE SAMPLE COUNT")
print("========================")

print("Kharif:", len(kharif_yield))
print("Rabi  :", len(rabi_yield))
print("Zaid  :", len(zaid_yield))


# ------------------------------------------------------------
# 6.3 - AVERAGE YIELD BY SEASON
# ------------------------------------------------------------

season_yield_summary = analysis_df.groupby(
    "Season"
)["Yield_Tonnes_Ha"].agg(
    ["count", "mean", "median", "min", "max"]
).round(2)

print("\nSEASON-WISE YIELD SUMMARY")
print("=========================")

display(season_yield_summary)


# ------------------------------------------------------------
# 6.4 - KRUSKAL-WALLIS TEST
# ------------------------------------------------------------

statistic, p_value = kruskal(
    kharif_yield,
    rabi_yield,
    zaid_yield
)


# ------------------------------------------------------------
# 6.5 - TEST RESULTS
# ------------------------------------------------------------

print("\nKRUSKAL-WALLIS TEST")
print("===================")

print("Test Statistic :", round(statistic, 4))
print("P-value        :", round(p_value, 6))


# ------------------------------------------------------------
# 6.6 - INTERPRETATION
# ------------------------------------------------------------

alpha = 0.05

print("\nSTATISTICAL INTERPRETATION")
print("==========================")

if p_value < alpha:
    print("Result: The seasonal yield differences are statistically significant.")
    print("Decision: Reject the null hypothesis (H0).")
else:
    print("Result: No statistically significant difference was detected among seasons.")
    print("Decision: Fail to reject the null hypothesis (H0).")


# ------------------------------------------------------------
# 6.7 - HYPOTHESES
# ------------------------------------------------------------

print("\nHYPOTHESES")
print("==========")

print("H0: Yield distributions are the same across Kharif, Rabi and Zaid.")
print("H1: At least one season has a different yield distribution.")


# ------------------------------------------------------------
# 6.8 - VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

analysis_df.boxplot(
    column="Yield_Tonnes_Ha",
    by="Season"
)

plt.title("Yield Distribution Across Agricultural Seasons")
plt.suptitle("")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# FINAL MESSAGE
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("STEP 6 - STATISTICAL TESTING COMPLETED SUCCESSFULLY")
print("=" * 60)
# ============================================================
# STEP 7 - SAVE FINAL ANALYSIS OUTPUTS
# ============================================================

import os

# ------------------------------------------------------------
# 7.1 - CREATE OUTPUT FOLDERS
# ------------------------------------------------------------

os.makedirs("/content/outputs/charts", exist_ok=True)
os.makedirs("/content/outputs/tables", exist_ok=True)

print("Output folders created successfully.")


# ------------------------------------------------------------
# 7.2 - SAVE ANALYSIS TABLES
# ------------------------------------------------------------

# Environmental analysis
environmental_summary.to_csv(
    "/content/outputs/tables/environmental_summary.csv"
)

# Nutrient analysis
nutrient_summary.to_csv(
    "/content/outputs/tables/nutrient_summary.csv"
)

# Water analysis
water_summary.to_csv(
    "/content/outputs/tables/water_summary.csv"
)

# Irrigation analysis
irrigation_summary.to_csv(
    "/content/outputs/tables/irrigation_summary.csv"
)

# Economic analysis
economic_summary.to_csv(
    "/content/outputs/tables/economic_summary.csv"
)

# Crop economic analysis
crop_economic.to_csv(
    "/content/outputs/tables/crop_economic.csv"
)

# Correlation matrix
correlation_matrix.to_csv(
    "/content/outputs/tables/correlation_matrix.csv"
)

# Yield correlations
yield_correlations.to_csv(
    "/content/outputs/tables/yield_correlations.csv"
)

# Crop yield summary
crop_yield_summary.to_csv(
    "/content/outputs/tables/crop_yield_summary.csv"
)

# Statistical test result
statistical_result = pd.DataFrame({
    "Test": ["Kruskal-Wallis"],
    "Statistic": [statistic],
    "P_Value": [p_value],
    "Significance_Level": [0.05],
    "Result": [
        "Significant"
        if p_value < 0.05
        else "Not Significant"
    ]
})

statistical_result.to_csv(
    "/content/outputs/tables/statistical_test_result.csv",
    index=False
)


# ------------------------------------------------------------
# 7.3 - SAVE IMPORTANT CHARTS
# ------------------------------------------------------------

# Seasonal Yield Chart
season_yield = analysis_df.groupby(
    "Season"
)["Yield_Tonnes_Ha"].mean()

plt.figure(figsize=(8, 5))

season_yield.plot(kind="bar")

plt.title("Average Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=0)

plt.tight_layout()

plt.savefig(
    "/content/outputs/charts/average_yield_by_season.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ------------------------------------------------------------
# Crop Yield Chart
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

crop_yield_summary["mean"].sort_values().plot(
    kind="barh"
)

plt.title("Average Yield by Crop")
plt.xlabel("Average Yield (Tonnes/Ha)")
plt.ylabel("Crop")

plt.tight_layout()

plt.savefig(
    "/content/outputs/charts/average_yield_by_crop.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ------------------------------------------------------------
# Water Efficiency Chart
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

water_summary[
    "Average_Water_Efficiency"
].sort_values().plot(
    kind="barh"
)

plt.title("Water Efficiency by Season")
plt.xlabel("Water Efficiency (t per 1000 m³)")
plt.ylabel("Season")

plt.tight_layout()

plt.savefig(
    "/content/outputs/charts/water_efficiency_by_season.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ------------------------------------------------------------
# Economic Performance Chart
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

economic_summary[
    [
        "Average_Revenue",
        "Average_Cost",
        "Average_Profit"
    ]
].plot(
    kind="bar"
)

plt.title("Economic Performance by Season")
plt.xlabel("Season")
plt.ylabel("Average Value (INR)")
plt.xticks(rotation=0)

plt.tight_layout()

plt.savefig(
    "/content/outputs/charts/economic_performance_by_season.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ------------------------------------------------------------
# Correlation Heatmap
# ------------------------------------------------------------

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix of Agricultural Variables")

plt.tight_layout()

plt.savefig(
    "/content/outputs/charts/correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ------------------------------------------------------------
# Statistical Test Boxplot
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

analysis_df.boxplot(
    column="Yield_Tonnes_Ha",
    by="Season"
)

plt.title("Yield Distribution Across Agricultural Seasons")
plt.suptitle("")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")

plt.tight_layout()

plt.savefig(
    "/content/outputs/charts/seasonal_yield_boxplot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ------------------------------------------------------------
# 7.4 - DISPLAY SAVED FILES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("STEP 7 COMPLETED")
print("=" * 60)

print("\nTables saved:")
for file in sorted(os.listdir("/content/outputs/tables")):
    print("✔", file)

print("\nCharts saved:")
for file in sorted(os.listdir("/content/outputs/charts")):
    print("✔", file)

print("\nAll analysis outputs have been saved successfully.")
# Final Analysis

## Key Findings

##The analysis was performed to understand the effect of seasons, crops, environmental conditions, resource usage, irrigation methods, and economic factors on agricultural performance.

#The major observations obtained from the analysis are presented through tables, charts, correlation analysis, and statistical testing.

## Seasonal Performance

#Season-wise agricultural performance was compared using yield, production, profit, water usage, water efficiency, and disease or pest risk.

## Crop Performance

#Different crops were compared across seasons to understand variations in average yield and overall agricultural performance.

## Environmental and Resource Analysis

#Environmental conditions and resource usage were analyzed to understand their statistical relationship with agricultural yield and other performance indicators.

## Statistical Analysis

#Correlation analysis and the Kruskal-Wallis test were used to support the analysis with statistical evidence.

## Conclusion

#The project provides a data-driven overview of seasonal agricultural performance. The findings can help in understanding seasonal variations, comparing crop performance, evaluating resource usage, and supporting better agricultural planning.
```
